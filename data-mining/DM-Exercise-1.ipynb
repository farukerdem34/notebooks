# Dataset Generation

The NumPy and pandas libraries required for data generation and analysis are imported. The tools used in this step form the basis of the subsequent calculations.

```python
import numpy as np
import pandas as pd
```

# Dataset Generation

A dataset of 10 rows and 5 columns consisting of random integer values is generated. This data is used in subsequent statistical and distance calculations.

```python
f1 = np.random.randint(1,10,10)
f2 = np.random.randint(0,5,10)
f3 = np.random.randint(5,25,10)
f4 = np.random.randint(0,5,10)
f5 = np.random.randint(1,15,10)
data = np.column_stack((f1,f2,f3,f4,f5))
data
```

# Table View

The generated dataset is displayed as a table in the form of a pandas DataFrame. This way, the structure and values of the data can be easily inspected.

```python
pd.DataFrame(data).round(2)
```

# Mean, Standard Deviation and Variance Calculations With Numpy

Using NumPy, the mean, standard deviation, and variance are calculated for each feature. These statistics summarize the central tendency and spread of the data.

```python
means = np.mean(data, axis=0)
stds = np.std(data, axis=0)
variances = np.var(data, axis=0)
```

# Printing the Statistics

The calculated mean, standard deviation, and variance values are printed for each feature. The results summarize the central tendency and spread of the data.

```python
for i in range(0,len(means)):
  print(f"Mean of feature {i+1} is {means[i]}")
  print(f"Standart Deviasion of feature {i+1} is {stds[i]}")
  print(f"Variance of feature {i+1} is {variances[i]}")
  print("\n---\n")
```

# Data Centralization

The data is centered by subtracting the mean of each feature from itself. The centered data is used in covariance and distance calculations.

```python
data-means
```

# Transpose

The transpose of the data matrix is taken, swapping rows and columns. This operation is necessary for matrix multiplication and covariance calculations.

```python
pd.DataFrame(data.T).round(2)
```

# Matrix Multiplipication

The transpose of the data matrix is multiplied by itself to obtain sums of squares and cross-products. This operation forms the basis of the covariance calculation.

```python
np.matmul(data.T,data)
```

# Covariance Calculation With Numpy

The covariance matrix of the data is calculated using NumPy's cov function. This matrix shows the linear relationship between the variables.

```python
cov_matrix = np.cov(data,rowvar=False)
pd.DataFrame(cov_matrix).round(2)
```

## Or

The same covariance matrix is also calculated step by step manually using the centered data. This way, the NumPy result is verified.

```python
centr_data = data - means
print("centered data")
display(pd.DataFrame(centr_data).round(2))
print("transposed centered data")
display(pd.DataFrame(centr_data.T).round(2))
print("matrix multiplication")
display(pd.DataFrame(np.matmul(centr_data.T,centr_data)).round(2))
manual_cov = np.matmul(centr_data.T,centr_data)/(len(data)-1)
print("manual covariance")
display(pd.DataFrame(manual_cov).round(2))
```

# Inverted Covariance for Mahalanobis Distance Calculation With Numpy

The inverse of the covariance matrix is taken. This inverse matrix is used in the calculation of the Mahalanobis distance.

```python
inv_cov_matrix = np.linalg.inv(cov_matrix)
pd.DataFrame(inv_cov_matrix).round(2)
```

# Euclidean, Cosine, Manhattan, Mahalanobis Distance Calculations With Scipy

Using SciPy, the Euclidean, cosine, Manhattan, and Mahalanobis distances are calculated between the selected sample pairs. The results of the different distance measures are compared.

```python
from scipy.spatial import distance

pairs = [(0,1),(2,3),(4,5)]

for i,j in pairs:
    u = data[i]
    v = data[j]

    print(f"\nSample {i} and Sample {j}:")
    print(f"  Euclidean: {distance.euclidean(u, v):.2f}")
    print(f"  Cosine: {distance.cosine(u, v):.4f}")
    print(f"  Manhattan: {distance.cityblock(u, v):.2f}")
    print(f"  Mahalanobis: {distance.mahalanobis(u, v, inv_cov_matrix):.2f}")
```
